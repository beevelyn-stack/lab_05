```python
import json
import requests

item_urls = {
    "https://www.loc.gov/resource/cph.3f05183/",
    "https://www.loc.gov/resource/fsa.8d24709/",
    "https://www.loc.gov/resource/highsm.64003/",
}

for number, url in enumerate(item_urls, start=1):
    response = requests.get(url, params={"fo": "json"}, timeout=30)
    print(f"Status code: {response.status_code}")
    response.raise_for_status()
    item_data = response.json()

    filename = f"item_{number}.json"
    with open(filename, "w", encoding="utf-8") as file:
        json.dump(item_data, file, ensure_ascii=False, indent=2)

    print(f"Saved {filename}")
```

```python
import requests

base_url = "https://www.loc.gov/search"
query = input("Enter Library of Congress search query: ")

parameters = {
    "q": query,
    "fo": "json",
}

response = requests.get(base_url, params=parameters)
print(f"Status code: {response.status_code}")
response.raise_for_status()

print(response.json())
```
